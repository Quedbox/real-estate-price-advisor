## EDA DF preparing

```python
df_eda = df.copy()

df_eda["log_price"] = np.log1p(df_eda["price"])
df_eda["log_area"] = np.log1p(df_eda["area"])
df_eda["price_per_m2"] = df_eda["price"] / df_eda["area"]

df_eda.head()
```

## Distribution profile

Price and area have long right tails. The table below compares quantiles and sample skewness/excess kurtosis; the deterministic sample keeps the higher-moment calculations responsive on the full dataset.

```python
profile_columns = ["price", "log_price", "area", "price_per_m2"]
profile_sample = df_eda[profile_columns].sample(
    n=min(100_000, len(df_eda)), random_state=42
)
distribution_profile = pd.DataFrame(
    {
        column: {
            "count": int(df_eda[column].notna().sum()),
            "median": df_eda[column].median(),
            "p95": df_eda[column].quantile(0.95),
            "p99": df_eda[column].quantile(0.99),
            "sample_skewness": skew(profile_sample[column], bias=False, nan_policy="omit"),
            "sample_excess_kurtosis": kurtosis(
                profile_sample[column], bias=False, nan_policy="omit"
            ),
        }
        for column in profile_columns
    }
).T
distribution_profile
```

## Target distribution: price

```python
price_quantiles = df_eda["price"].quantile([0.01, 0.05, 0.5, 0.95, 0.99, 0.999])
price_quantiles
```

```python
price_p99 = df_eda["price"].quantile(0.99)

fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df_eda.loc[df_eda["price"] <= price_p99, "price"], bins=50, color="#2f6690")
ax.set_title("Listing price distribution (up to the 99th percentile)")
ax.set_xlabel("Price (RUB)")
ax.set_ylabel("Listings")
ax.ticklabel_format(style="plain", axis="x")
fig.tight_layout()
PRICE_DISTRIBUTION_PATH = FIGURES_PATH / "price_distribution.png"
fig.savefig(PRICE_DISTRIBUTION_PATH, dpi=240, bbox_inches="tight")
plt.show()
plt.close(fig)
```

## Numeric correlation

```python
numeric_columns = [
    "price",
    "log_price",
    "area",
    "log_area",
    "kitchen_area",
    "rooms",
    "floor",
    "floors",
    "floor_ratio",
    "geo_lat",
    "geo_lon",
]

existing_numeric_columns = [
    column for column in numeric_columns
    if column in df_eda.columns
]

corr = df_eda[existing_numeric_columns].corr(numeric_only=True)

corr["price"].sort_values(ascending=False)
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(corr, vmin=-1, vmax=1, cmap="coolwarm")
fig.colorbar(im, ax=ax)

ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.index)))
ax.set_xticklabels(corr.columns, rotation=90)
ax.set_yticklabels(corr.index)
ax.set_title("Numeric feature correlation matrix")

for i in range(len(corr.index)):
    for j in range(len(corr.columns)):
        value = corr.iloc[i, j]
        ax.text(j, i, f"{value:.2f}", ha="center", va="center", fontsize=8)

fig.tight_layout()
CORRELATION_FIGURE_PATH = FIGURES_PATH / "correlation_matrix.png"
fig.savefig(CORRELATION_FIGURE_PATH, dpi=240, bbox_inches="tight")
plt.show()
plt.close(fig)
```
